```python
# ================================
# DATA CLEANING
# ================================

# 1. Xóa các dòng trùng lặp
df_clean = df_clean.drop_duplicates()

# 2. Xử lý area bất thường
df_clean.loc[df_clean["area"] > 1000, "area"] = None

# 3. Xử lý year_built bất thường
df_clean.loc[df_clean["year_built"] < 1800, "year_built"] = None

# 4. Xử lý floor âm
# Xem tầng âm là dữ liệu không chuẩn cho mô hình
df_clean.loc[df_clean["floor"] < 0, "floor"] = None

# 5. Xử lý các cột điểm đánh giá
rating_cols = [
    "index",
    "environment",
    "quality_of_living",
    "safety",
    "transport",
    "services",
    "relax"
]

# Giới hạn điểm trong khoảng hợp lý
for col in rating_cols:
    df_clean.loc[
        (df_clean[col] < 0) | (df_clean[col] > 10),
        col
    ] = None

# 6. Kiểm tra kết quả
print("Số dòng ban đầu:", len(df))
print("Số dòng sau cleaning:", len(df_clean))
print("Số dòng trùng lặp còn lại:", df_clean.duplicated().sum())

print("\nMissing values:")
print(
    df_clean.isnull()
    .sum()
    .sort_values(ascending=False)
)
```

```python
# ================================
# HANDLE MISSING VALUES
# ================================

# 1. Cột số: điền bằng median
numeric_cols = df_clean.select_dtypes(include="number").columns

for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())


# 2. Cột dạng text: điền bằng "Unknown"
categorical_cols = df_clean.select_dtypes(include="object").columns

for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna("Unknown")


# 3. Kiểm tra lại missing
print("Tổng số missing còn lại:")
print(df_clean.isnull().sum().sum())
```

```python

from pyspark.sql import functions as F

# Chọn các cột dùng để dự đoán giá
feature_cols = [
    "area",
    "rooms",
    "year_built",
    "total_floors",
    "floor",
    "lift",
    "balkonies",
    "loggia",
    "cellar",
    "district",
    "type",
    "construction_type",
    "condition",
    "orientation"
]

# Giữ các cột cần thiết và loại bản ghi thiếu dữ liệu
model_df = spark_df.select(
    ["price"] + feature_cols
).dropna()

print("Số dòng dùng để huấn luyện:", model_df.count())
print("Số đặc trưng:", len(feature_cols))

model_df.show(5, truncate=False)
```

```python

from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler

# Tự động nhận diện cột chữ và cột số
categorical_cols = [
    c for c, dtype in model_df.dtypes
    if dtype == "string"
]

numeric_cols = [
    c for c, dtype in model_df.dtypes
    if c != "price" and dtype != "string"
]

print("Các cột chữ:", categorical_cols)
print("Các cột số:", numeric_cols)

# Chuyển cột chữ thành chỉ số số học
indexers = [
    StringIndexer(
        inputCol=c,
        outputCol=f"{c}_idx",
        handleInvalid="keep"
    )
    for c in categorical_cols
]

# Chuyển chỉ số thành vector mã hóa
encoder = OneHotEncoder(
    inputCols=[f"{c}_idx" for c in categorical_cols],
    outputCols=[f"{c}_vec" for c in categorical_cols],
    handleInvalid="keep"
)

# Gộp các đặc trưng thành một vector
assembler = VectorAssembler(
    inputCols=numeric_cols + [f"{c}_vec" for c in categorical_cols],
    outputCol="features",
    handleInvalid="keep"
)

# Tạo pipeline xử lý
stages = indexers + [encoder, assembler]
pipeline = Pipeline(stages=stages)

# Fit và biến đổi dữ liệu để kiểm tra
pipeline_model = pipeline.fit(model_df)
prepared_df = pipeline_model.transform(model_df)

prepared_df.select("price", "features").show(5, truncate=False)

print("Số dòng:", prepared_df.count())
print("Đã mã hóa và tạo đặc trưng thành công!")
```

```python

from pyspark.ml.regression import RandomForestRegressor

# 1. Chia dữ liệu: 80% huấn luyện, 20% kiểm tra
train_df, test_df = prepared_df.randomSplit(
    [0.8, 0.2],
    seed=42
)

print("Số dòng huấn luyện:", train_df.count())
print("Số dòng kiểm tra:", test_df.count())

# 2. Tạo mô hình Random Forest
rf = RandomForestRegressor(
    featuresCol="features",
    labelCol="price",
    numTrees=50,
    maxDepth=10,
    seed=42
)

# 3. Huấn luyện mô hình
rf_model = rf.fit(train_df)

print("Huấn luyện mô hình thành công!")

# 4. Dự đoán trên tập kiểm tra
predictions = rf_model.transform(test_df)

predictions.select(
    "price",
    "prediction"
).show(10, truncate=False)
```

```python

from pyspark.ml.evaluation import RegressionEvaluator

# Tính các chỉ số đánh giá
metrics = {}

for metric_name in ["mae", "rmse", "r2"]:
    evaluator = RegressionEvaluator(
        labelCol="price",
        predictionCol="prediction",
        metricName=metric_name
    )
    metrics[metric_name.upper()] = evaluator.evaluate(predictions)

# Hiển thị kết quả
print("===== KẾT QUẢ ĐÁNH GIÁ RANDOM FOREST =====")
print(f"MAE  : {metrics['MAE']:,.2f}")
print(f"RMSE : {metrics['RMSE']:,.2f}")
print(f"R²   : {metrics['R2']:.4f}")
```

```python

from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler
from pyspark.ml.regression import RandomForestRegressor
from pyspark.ml.evaluation import RegressionEvaluator

# 1. Chia dữ liệu TRƯỚC khi học cách mã hóa
train_df, test_df = model_df.randomSplit([0.8, 0.2], seed=42)

# 2. Tự động xác định các cột chữ và cột số
categorical_cols = [
    c for c, dtype in model_df.dtypes
    if dtype == "string"
]

numeric_cols = [
    c for c, dtype in model_df.dtypes
    if c != "price" and dtype != "string"
]

# 3. Xây dựng quy trình mã hóa tự động
indexers = [
    StringIndexer(
        inputCol=c,
        outputCol=f"{c}_idx",
        handleInvalid="keep"
    )
    for c in categorical_cols
]

encoder = OneHotEncoder(
    inputCols=[f"{c}_idx" for c in categorical_cols],
    outputCols=[f"{c}_vec" for c in categorical_cols],
    handleInvalid="keep"
)

assembler = VectorAssembler(
    inputCols=numeric_cols + [f"{c}_vec" for c in categorical_cols],
    outputCol="features",
    handleInvalid="keep"
)

# 4. Tạo mô hình Random Forest
rf = RandomForestRegressor(
    featuresCol="features",
    labelCol="price",
    numTrees=50,
    maxDepth=10,
    seed=42
)

pipeline = Pipeline(stages=indexers + [encoder, assembler, rf])

# 5. Chỉ học từ tập train
model = pipeline.fit(train_df)

# 6. Dự đoán trên tập test
predictions = model.transform(test_df)

# 7. Đánh giá mô hình
for metric in ["mae", "rmse", "r2"]:
    evaluator = RegressionEvaluator(
        labelCol="price",
        predictionCol="prediction",
        metricName=metric
    )
    print(f"{metric.upper()}: {evaluator.evaluate(predictions):,.4f}")
```

```python

from pyspark.ml.regression import DecisionTreeRegressor
from pyspark.ml import Pipeline
from pyspark.ml.evaluation import RegressionEvaluator

# Tạo Decision Tree
dt = DecisionTreeRegressor(
    featuresCol="features",
    labelCol="price",
    maxDepth=10,
    seed=42
)

# Tạo pipeline với cùng bước xử lý dữ liệu
dt_pipeline = Pipeline(
    stages=indexers + [encoder, assembler, dt]
)

# Huấn luyện trên cùng tập train
dt_model = dt_pipeline.fit(train_df)

# Dự đoán trên cùng tập test
dt_predictions = dt_model.transform(test_df)

# Đánh giá
for metric in ["mae", "rmse", "r2"]:
    evaluator = RegressionEvaluator(
        labelCol="price",
        predictionCol="prediction",
        metricName=metric
    )
    value = evaluator.evaluate(dt_predictions)
    print(f"Decision Tree - {metric.upper()}: {value:,.4f}")
```

```python
# Đọc dữ liệu đã làm sạch bằng PySpark
spark_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option("encoding", "UTF-8")
    .csv("../data/processed/real_estate_clean.csv")
)

# Kiểm tra dữ liệu
print("Số dòng:", spark_df.count())
print("Số cột:", len(spark_df.columns))

# Hiển thị 5 dòng đầu tiên
spark_df.show(5, truncate=False)
```

```python
from pyspark.sql import functions as F

by_district = (
    spark_df
    .groupBy("district")
    .agg(
        F.count("*").alias("count"),
        F.avg("price").alias("avg_price"),
        F.min("price").alias("min_price"),
        F.max("price").alias("max_price")
    )
    .orderBy(F.desc("avg_price"))
)

by_district.show(10, truncate=False)
```

```python
from pyspark.sql import functions as F

# 1. Giá theo khu vực
by_district = (
    spark_df.groupBy("district")
    .agg(
        F.count("*").alias("count"),
        F.avg("price").alias("avg_price"),
        F.min("price").alias("min_price"),
        F.max("price").alias("max_price")
    )
    .orderBy(F.desc("avg_price"))
)

# 2. Giá theo loại bất động sản
by_type = (
    spark_df.groupBy("type")
    .agg(
        F.count("*").alias("count"),
        F.avg("price").alias("avg_price")
    )
    .orderBy(F.desc("avg_price"))
)

# 3. Giá theo số phòng
by_rooms = (
    spark_df.groupBy("rooms")
    .agg(
        F.count("*").alias("count"),
        F.avg("price").alias("avg_price")
    )
    .orderBy("rooms")
)

# 4. Giá theo nhóm diện tích
by_area = (
    spark_df.withColumn(
        "area_group",
        F.when(F.col("area") < 50, "Dưới 50")
         .when(F.col("area") < 100, "50-99")
         .when(F.col("area") < 150, "100-149")
         .otherwise("Từ 150 trở lên")
    )
    .groupBy("area_group")
    .agg(
        F.count("*").alias("count"),
        F.avg("price").alias("avg_price")
    )
)

# Kiểm tra số nhóm và hiển thị kết quả
print("Số khu vực:", by_district.count())
print("Số loại bất động sản:", by_type.count())
print("Số nhóm số phòng:", by_rooms.count())
print("Số nhóm diện tích:", by_area.count())

print("\nGiá theo loại bất động sản:")
by_type.show(truncate=False)

print("\nGiá theo số phòng:")
by_rooms.show(truncate=False)

print("\nGiá theo nhóm diện tích:")
by_area.show(truncate=False)
```
